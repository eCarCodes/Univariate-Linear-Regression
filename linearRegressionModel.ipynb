```python
import numpy as np
import matplotlib.pyplot as plt
import csv
```

```python
def compute_model_output(x, w, b):
    """ 
    Computes the prediction of a linear model (we do this for every input x)
    Args:
        x (ndarray(x)): data, m examples
        w, b (scalar): we usually retrive from gradient descent
    Returns:
        f_wb(nparray(m)): model prediction
    """
    m = x.shape[0]              # Grabs the first dimension from the tupled elements of the array: from (30,) grab index 0 -> expected: 30 elements
    f_wb = np.zeros(m)          # fill all array slots with "m" many 0's
    for i in range(m):
        f_wb[i] = w * x[i] + b  # calculate regression for each input
    return f_wb                 # return an array of linear regression per data point
```

```python
# Load the dataset into a numpy array
# We create a 2D list using our loaded CSV data, then turn it into a numpy array for easy calculations
file_path = "C:\\Users\\eacar\\OneDrive\\Desktop\\PythonCode\\machineLearning\\Univariate_LR\\data\\Experience-Salary.csv" # ENTER YOUR FILE PATH HERE
with open(file_path, "r") as file:
    content = csv.reader(file)
    new_list = []
    # Append each rows elements into a nested list
    for row in content:
        add_list = []
        for column in row:
            add_list.append(float(column))
        new_list.append(add_list)
    # Then, turn it into a numpy array for calculations
    inputArray = np.array(new_list)

x = inputArray[:,0]         # Grab the 1st element (months worked) in each row
y = inputArray[:,1] * 1000  # Grab the 2nd element (salary in thousands) in each row
print(f"x values: {x}")
print(f"\nTrue y values: {y}")
```

```python
def loss_function(w, b):
    """
    Measures how wrong the model is (squared difference) between its predictions and the actual y outputs

    Returns:
        Gradient descent returns the updated weights of w and b that best accurately fits our data.
    Notes:
        Squaring the error penalizes the models performance greatly, influencing a stronger prediction
    """
    m = x.shape[0]
    total_sum = 0
    for i in range(m):
        f_wb = w * x[i] + b
        cost = (f_wb - y[i])**2
        total_sum = total_sum + cost
    loss = (1.0/(2.0 * m)) * total_sum
    return loss
```

```python
def compute_gradient():
    """
    Computes a prediction to calculate the weights of w and b automatically 
    to minimize the cost function J, and to accurately plot against our data.

    Returns:
        Gradient descent returns the updated weights of w and b that best accurately fits our data.
    Notes:
        The learning rate (alpha) adjusts the "step size" for gradient descent, usually set between 0 - 1.
        We calculate the partial derivative of cost function J to w and b to determine the rate of change
        relative to each parameter so we know which way to adjust them.
    """
    m = x.shape[0]
    alpha = 0.001
    iterations = 4000
    # It's typical to set parameters to 0
    w = 0
    b = 0
    # Perform the gradient, update the weights, and loop until convergence
    for iterate in range(iterations):
        total_sum_w = 0
        total_sum_b = 0
        for i in range(m):
            f_wb = w * x[i] + b
            cost_w = (f_wb - y[i]) * x[i]
            cost_b = (f_wb - y[i])
            total_sum_w  = total_sum_w + cost_w
            total_sum_b  = total_sum_b + cost_b
        derivative_w = (1 / m) * total_sum_w
        derivative_b = (1 / m) * total_sum_b
        temp_w = w - alpha * derivative_w
        temp_b = b - alpha * derivative_b
        w = temp_w
        b = temp_b
    return w,b
```

```python
w,b = compute_gradient()
print(f"Calculated 'w: {w}")
print(f"Calculated 'b: {b}")
```

```python
# Set model parameters and calculate predictions
# We commonly start by setting our weights to 0 to allow for our gradient descent to adjust accordingly.

w = 0 # CALCULATE
b = 0 # CALCULATE

f_wb = compute_model_output(x, w, b)
print("Model Predictions:", f_wb)
```

```python
# Check the metrics
loss = loss_function(w, b)
print(f"Loss Error: {loss:.2f}") # rounded to 2 decimal places
```

```python
# Plot actual data and model predictions graphically
plt.scatter(x, y, marker='x', color='red', label='Actual data')
plt.plot(x, f_wb, color='blue', label='Model predictions')

plt.xlabel("Months Worked (x)")
plt.ylabel("Salary (y)")
plt.title("Linear Regression: Predictions vs Actual Data")
plt.legend()
plt.grid(True)
plt.show()

# .scatter(): (inputs (x), actual outputs (y), marker, color, label)
# The key is that .plot() takes your x-values and y-values, pairs them together, and connects the resulting points in order. 
# .plot() -> Creates coordinates (x inputs, your_fwb_predictions (y-hat), color, label)
```
